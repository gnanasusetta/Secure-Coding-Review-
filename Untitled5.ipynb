```python
# TASK 3: Secure Coding Review - Basic Security Audit

import ast

code = '''
import sqlite3
import os

def login(username, password):
    conn = sqlite3.connect("users.db")
    query = "SELECT * FROM users WHERE username = '" + username + "' AND password = '" + password + "'"
    result = conn.execute(query)
    return result.fetchone()

def run_command(user_input):
    os.system("ping " + user_input)

password = "admin123"
'''

print("=" * 65)
print("             SECURE CODING REVIEW")
print("=" * 65)

print("\nSelected Language : Python")
print("Audit Method      : Manual Inspection + Static Analysis\n")

tree = ast.parse(code)
findings = []

for node in ast.walk(tree):

    # Detect hardcoded passwords
    if isinstance(node, ast.Assign):
        for target in node.targets:
            if isinstance(target, ast.Name) and "password" in target.id.lower():
                findings.append(("HIGH", "Hardcoded password detected"))

    # Detect SQL string concatenation
    if isinstance(node, ast.BinOp) and isinstance(node.op, ast.Add):
        if isinstance(node.left, ast.Constant) and isinstance(node.left.value, str):
            if "SELECT" in node.left.value.upper():
                findings.append(("HIGH", "Possible SQL Injection vulnerability"))

    # Detect os.system usage
    if isinstance(node, ast.Call):
        if isinstance(node.func, ast.Attribute):
            if node.func.attr == "system":
                findings.append(("HIGH", "Potential OS Command Injection"))

print("SECURITY FINDINGS")
print("-" * 65)

if findings:
    for severity, issue in findings:
        print(f"[{severity}] {issue}")

print("\nRECOMMENDATIONS")
print("-" * 65)
print("1. Use parameterized SQL queries instead of string concatenation.")
print("2. Never hardcode passwords or secret keys in source code.")
print("3. Avoid os.system() with untrusted user input.")
print("4. Validate and sanitize all user-provided input.")
print("5. Use secure password hashing and environment variables for secrets.")

print("\nREMEDIATION")
print("-" * 65)
print("✓ Replace vulnerable SQL queries with prepared statements.")
print("✓ Store secrets securely using environment variables.")
print("✓ Use safer subprocess methods with validated arguments.")
print("✓ Perform regular static analysis and code reviews.")

print("=" * 65)
print("Secure Coding Review Completed!")
print("=" * 65)
```
